# MYELIN — Experimento 7: o ciclo real, linhas de base justas e o reflexo

Responde à revisão externa do Experimento 6 (`docs/resultados-exp6.md`, "Correção"):

- simula o **ciclo real** da decodificação especulativa (rascunho → verificação → token bônus, professor com um
  ciclo de atraso) sobre texto **gerado pelo próprio modelo**, com o custo corrigido;
- linhas de base justas: **busca de n-gramas** (Prompt Lookup Decoding) e **cache neural** corrigido;
- **controles** da memória: pares embaralhados, correção média, última correção;
- o **reflexo do MYELIN**: reconhecer a situação pelo estado oculto e repetir o que o verificador disse, sem rodar a
  cabeça.

Os hiperparâmetros de todos são escolhidos nos 7 textos de validação (`data/eval/`) e medidos nos 15 de teste.
